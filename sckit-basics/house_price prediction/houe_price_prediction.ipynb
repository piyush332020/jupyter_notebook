```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

df = pd.read_csv("train_house.csv")

x = df.drop(columns=["Price_in_Lakhs"])
y = df["Price_in_Lakhs"]

x = pd.get_dummies(x, drop_first=True)

print("Training Data Shape:", x.shape)

model = RandomForestRegressor(
    n_estimators=100,   
    random_state=42,
    n_jobs=-2          
)

print("Training the model...")
model.fit(x, y)
print("Model training completed!")

test_data = pd.read_csv("test_house.csv")


x_test = pd.get_dummies(test_data, drop_first=True)

x_test = x_test.reindex(columns=x.columns, fill_value=0)

# print("Making predictions...")
# for i in range(0,99):
#     prediction = model.predict(x_test.values)
#     print(f"Predictions {i}:")
# print(prediction)
```

```python
print("Making predictions...")

prediction = model.predict(x_test.values)
for i, pred in enumerate(prediction):
    print(f"House {i+1}: {pred}")
```

```python
print("Making diffrence...")
total=0;
prediction = model.predict(x_test.values)
for i, pred in enumerate(prediction):
    print(f"House diff {i}: {abs((y[i]-pred))}")
    total+=abs(y[i]-pred)
print(f"average diffrence in price are :{len(prediction)/total}")
```
