```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

df = pd.read_csv("train_house.csv")
# Remove ID column
df = df.drop(columns="ID")

# Features and Target
x = df.drop(columns=["Price_in_Lakhs", "Price_per_SqFt"])
y = df["Price_in_Lakhs"]

# Convert categorical columns into numerical columns
x = pd.get_dummies(x, drop_first=True)

print("Training Data Shape:", x.shape)

# Create and train the model
model = RandomForestRegressor(
    n_estimators=20,    # Fewer trees for faster training
    random_state=42,
    n_jobs=-1           # Use all CPU cores
)

print("Training the model...")
model.fit(x, y)
print("Model training completed!")

# -------------------- Test Data --------------------
test_data = pd.read_csv("test_house.csv")

# Remove ID column
test_data = test_data.drop(columns="ID")

# Convert categorical columns
x_test = pd.get_dummies(test_data, drop_first=True)

# Match test columns with training columns
x_test = x_test.reindex(columns=x.columns, fill_value=0)

print("Test Data Shape:", x_test.shape)

# Predict
print("Making predictions...")
prediction = model.predict(x_test)

print("Predictions:")
print(prediction)
```
