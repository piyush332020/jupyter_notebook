```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]


plt.bar(years,kohli ,label='kholi' ,color="green" )
plt.bar(years,sehwag ,label='shweag' ,color="red" )
plt.xlabel("year")
plt.ylabel("runs")
plt.legend()
plt.style.use('fivethirtyeight')
plt.grid(True)
```

```python
np.arange(len(years))
width=0.25
```

```python
x=np.arange(len(years))
width=0.25

plt.bar(x-width,kohli ,label='kholi',width=width,color="green" )
plt.bar(x,sehwag ,label='shweag',width=width,color="red" )


plt.xlabel("year")
plt.ylabel("runs")
plt.title("run comparision")
plt.xticks(x,years)
plt.legend()
plt.style.use('fivethirtyeight')
plt.grid(True)
```

```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs, color="skyblue")

# Add labels on top of bars
for i in range(len(players)):
    plt.text(i, runs[i] + 20, str(runs[i]), ha='center')
          # [player],[hieght], [display]

plt.ylabel("Runs")
plt.title("Runs Scored by Players")
plt.tight_layout()
plt.show()
```
