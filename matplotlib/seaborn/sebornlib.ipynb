```python
tips=sns.load_dataset("tips")
```

```python
tips.info()
```

```python
sns.scatterplot(x="total_bill",y="tip", label="cost", data=tips) #data=tips ....here tips is the dataset
plt.title("line plot")
plt.xlabel("tip")
plt.ylabel("cost")
plt.legend()
```
